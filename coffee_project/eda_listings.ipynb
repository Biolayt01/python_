```python
import numpy as np

n = 11
matrix = np.zeros((n, n), dtype=int)

# 1. Fill the main diagonal
np.fill_diagonal(matrix, 1)

# 2. Fill the anti-diagonal (by flipping horizontally)
np.fill_diagonal(np.fliplr(matrix), 1)

# 3. Fill the middle row and middle column
mid = n // 2
matrix[mid, :] = 1
matrix[:, mid] = 1

print(matrix)
```
