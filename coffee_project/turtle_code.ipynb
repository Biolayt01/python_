```python
from turtle import Turtle, Screen


turtle = Turtle()
screen = Screen()



#Set screen attributes
screen.bgcolor("maroon")
screen.setup(width=600, height=600)


# Set turtle attributes
print(turtle.turtlesize())

turtle.shape("turtle")
turtle.color("white")
turtle.forward(100)
turtle.left(90)
turtle.forward(100)
turtle.right(90)
turtle.forward(20)
turtle.left(90)
turtle.back(120)
turtle.left(90)
turtle.forward(120)
turtle.right(90)
turtle.forward(20)


screen.exitonclick()


```
